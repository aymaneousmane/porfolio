```python
import numpy as np
import ctypes

# Charger la bibliothèque MKL
mkl = ctypes.cdll.LoadLibrary("libmkl_rt.so")
```

```python
# cblas_ddot : Produit scalaire de deux vecteurs
def cblas_ddot(N, X, Y):
    X_ctypes = X.ctypes.data_as(ctypes.POINTER(ctypes.c_double))
    Y_ctypes = Y.ctypes.data_as(ctypes.POINTER(ctypes.c_double))
    mkl.cblas_ddot.restype = ctypes.c_double
    return mkl.cblas_ddot(N, X_ctypes, 1, Y_ctypes, 1)

# Exemple d'utilisation de cblas_ddot
X = np.array([1.0, 2.0, 3.0], dtype=np.float64)
Y = np.array([4.0, 5.0, 6.0], dtype=np.float64)
print("Produit scalaire:", cblas_ddot(3, X, Y))
```

```python
# Exemple d'utilisation de cblas_dgemm
A = np.arange(1000**2).reshape(1000, 1000)
B = np.random.rand(1000**2).reshape(1000, 1000)
```

- A : Première matrice de taille (M, K)
- B : Deuxième matrice de taille (K, N)
- M : Nombre de lignes de A et C
- N : Nombre de colonnes de B et C
- K : Nombre de colonnes de A et lignes de B

* A.ctypes.data_as(ctypes.POINTER(ctypes.c_double))
→ Convertit A en un pointeur C utilisable par cblas_dgemm.

Les paramètres :

- 101 = CblasRowMajor (les matrices sont stockées ligne par ligne)
- 111 = CblasNoTrans pour A et B (pas de transposition)
- M, N, K = Tailles des matrices
- 1.0 = Facteur multiplicatif α
- K, N = Nombre de colonnes des matrices A et B (respectivement lda, ldb)
- 0.0 = Facteur multiplicatif β
- N = Nombre de colonnes de C (ldc)
C=α⋅A⋅B+β⋅C

```python
# cblas_dgemm : Multiplication de matrices
def cblas_dgemm(A, B, M, N, K):
    A_ctypes = A.ctypes.data_as(ctypes.POINTER(ctypes.c_double))
    B_ctypes = B.ctypes.data_as(ctypes.POINTER(ctypes.c_double))
    C = np.zeros((M, N), dtype=np.float64)
    C_ctypes = C.ctypes.data_as(ctypes.POINTER(ctypes.c_double))
    
    mkl.cblas_dgemm.argtypes = [ctypes.c_int, ctypes.c_int, ctypes.c_int, 
                                ctypes.c_int, ctypes.c_int, ctypes.c_int,
                                ctypes.c_double, ctypes.POINTER(ctypes.c_double), ctypes.c_int,
                                ctypes.POINTER(ctypes.c_double), ctypes.c_int,
                                ctypes.c_double, ctypes.POINTER(ctypes.c_double), ctypes.c_int]
    
    mkl.cblas_dgemm(101, 111, 111, M, N, K, 1.0, A_ctypes, K, B_ctypes, N, 0.0, C_ctypes, N)
    return C

print("Multiplication matricielle:\n", cblas_dgemm(A, B, 1000, 1000, 1000))
```

```python
# cblas_dnrm2 : Norme Euclidienne
def cblas_dnrm2(N, X):
    X_ctypes = X.ctypes.data_as(ctypes.POINTER(ctypes.c_double))
    mkl.cblas_dnrm2.restype = ctypes.c_double
    return mkl.cblas_dnrm2(N, X_ctypes, 1)

# Exemple d'utilisation de cblas_dnrm2
X = np.array([3.0, 4.0], dtype=np.float64)
print("Norme Euclidienne:", cblas_dnrm2(2, X))

```

```python
# cblas_idamax : Indice de l'élément de plus grande valeur absolue
def cblas_idamax(N, X):
    X_ctypes = X.ctypes.data_as(ctypes.POINTER(ctypes.c_double))
    mkl.cblas_idamax.restype = ctypes.c_int
    return mkl.cblas_idamax(N, X_ctypes, 1)

# Exemple d'utilisation de cblas_idamax
X = np.array([1.0, -5.0, 3.0, -10.0, 7.0], dtype=np.float64)
print("Indice du max absolu:", cblas_idamax(5, X))
```
